## 6) Código funcional **[15 pts]**
Incluye tu implementación completa.

```python

import matplotlib.pyplot as plt

letras = ['A', 'B', 'C', 'D', 'E', 'F', 'G', 'H', 'I', 'J']

posiciones = {
    'F': 'A3',
    'L': 'C8',
    'S': 'B16',
    'P': 'D24',
    'B': 'H6',
    'R': 'I15',
    'M': 'G25'
}

def a_coordenadas(pos):
    fila = letras.index(pos[0]) + 1
    columna = int(pos[1:])
    return (columna, fila)

def a_posicion(coord):
    return letras[coord[1] - 1] + str(coord[0])

def distancia(p1, p2):
    return abs(p1[0] - p2[0]) + abs(p1[1] - p2[1])

def crear_aristas(componentes):
    nombres = list(componentes.keys())
    aristas = []
    for i in range(len(nombres)):
        for j in range(i + 1, len(nombres)):
            a = nombres[i]
            b = nombres[j]
            aristas.append((a, b, distancia(componentes[a], componentes[b])))
    return aristas

def buscar(padre, nodo):
    while padre[nodo] != nodo:
        nodo = padre[nodo]
    return nodo

def kruskal(nodos, aristas, mostrar=False):
    padre = {}
    for n in nodos:
        padre[n] = n

    mst = []
    total = 0

    for a, b, peso in sorted(aristas, key=lambda x: x[2]):
        raiz_a = buscar(padre, a)
        raiz_b = buscar(padre, b)
        if raiz_a != raiz_b:
            padre[raiz_b] = raiz_a
            mst.append((a, b, peso))
            total += peso
            if mostrar:
                print(f"Se elige {a} - {b} = {peso}")
            if len(mst) == len(nodos) - 1:
                break
        elif mostrar:
            print(f"Se descarta {a} - {b} = {peso} (forma un ciclo)")

    return mst, total

def dibujar(componentes, cables=None, titulo="", ax=None):
    solo = ax is None
    if solo:
        fig, ax = plt.subplots(figsize=(14, 6))

    ax.set_xlim(0.5, 30.5)
    ax.set_ylim(10.5, 0.3)
    ax.set_xticks(range(1, 31))
    ax.set_yticks(range(1, 11))
    ax.set_yticklabels(letras)
    ax.grid(True, alpha=0.3)
    ax.set_xlabel("Columnas")
    ax.set_ylabel("Filas")
    ax.set_title(titulo)

    if cables:
        for a, b, peso in cables:
            c1, f1 = componentes[a]
            c2, f2 = componentes[b]
            ax.plot([c1, c2, c2], [f1, f1, f2], color='black', linewidth=1, zorder=1)
            if c1 == c2 or f1 == f2:
                lx = (c1 + c2) / 2
                ly = (f1 + f2) / 2 - 0.3
            else:
                lx = c2
                ly = f1 - 0.3
            ax.text(lx, ly, str(peso), color='red', fontsize=9, ha='center', va='center', zorder=3)

    for i, (nombre, (c, f)) in enumerate(componentes.items()):
        ax.plot(c, f, 'o', color='C' + str(i), markersize=12, zorder=2)
        ax.text(c, f - 0.3, nombre, fontweight='bold', ha='center', va='bottom', zorder=3)

    if solo:
        plt.show()

def mostrar_resultado(mst, total, n):
    print("\tRESULTADO")
    for i, (a, b, peso) in enumerate(mst):
        print(f"Cable {i + 1}: {a} - {b} = {peso}")
    print(f"Longitud total del cable: {total} unidades")
    print(f"Cables: {len(mst)} de {n - 1}")

def menu():
    componentes = {}
    for nombre in posiciones:
        componentes[nombre] = a_coordenadas(posiciones[nombre])

    nodos = list(componentes.keys())
    print("Bienvenido se usara un menu con el fin de visualizar y ordenar las distintas herramientas que le dara a una protoboard.")
    while True:
        print("\tMENU")
        print("1. Ver protoboard inicial.")
        print("2. Ver las distancias entre componentes.")
        print("3. Ver todas las posibles conexiones.")
        print("4. Calcular Árbol de Expansión Mínima (Kruskal).")
        print("5. Ver protoboard con el resultado del MST.")
        print("6. Mover el Buzzer a J20 y recalcularlo.")
        print("0. Salir")

        try:
            opcion = int(input("Elige una opción: "))
        except ValueError:
            print("Escribe un número entero que este dentro del rango de las opsiones porfavor.")
            continue

        if opcion == 1:
            dibujar(componentes, titulo="Protoboard - Distribución Inicial")

        elif opcion == 2:
            aristas = crear_aristas(componentes)
            print("\nVértices:")
            for nombre, coord in componentes.items():
                print(f"{nombre} = {a_posicion(coord)} = {coord}")
            print("\nDistancias:")
            for a, b, peso in aristas:
                ca = componentes[a]
                cb = componentes[b]
                print(f"{a} {ca} - {b} {cb}: |{ca[0]} - {cb[0]}| + |{ca[1]} - {cb[1]}| = {peso}")

        elif opcion == 3:
            aristas = crear_aristas(componentes)
            print()
            for a, b, peso in aristas:
                print(f"{a} -------- {peso} -------- {b}")
            print(f"\nTotal de conexiones posibles: {len(aristas)}")

        elif opcion == 4:
            aristas = crear_aristas(componentes)
            print("\nConexiones ordenadas de menor a mayor peso:")
            for a, b, peso in sorted(aristas, key=lambda x: x[2]):
                print(f"{a} - {b} = {peso}")
            print()
            mst, total = kruskal(nodos, aristas, True)
            print()
            mostrar_resultado(mst, total, len(nodos))

        elif opcion == 5:
            aristas = crear_aristas(componentes)
            mst, total = kruskal(nodos, aristas)
            mostrar_resultado(mst, total, len(nodos))
            fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(22, 6))
            dibujar(componentes, titulo="Protoboard - Distribución Original", ax=ax1)
            dibujar(componentes, cables=mst, titulo="Protoboard - Distribución Original (MST)", ax=ax2)
            plt.show()

        elif opcion == 6:
            aristas = crear_aristas(componentes)
            mst, total = kruskal(nodos, aristas)

            componentes_nuevos = componentes.copy()
            componentes_nuevos['B'] = a_coordenadas('J20')

            aristas_nuevas = crear_aristas(componentes_nuevos)
            mst_nuevo, total_nuevo = kruskal(nodos, aristas_nuevas)

            print(f"\nMST original Longitud total: {total}")
            print(f"MST después del cambio Longitud total: {total_nuevo}\n")
            mostrar_resultado(mst_nuevo, total_nuevo, len(nodos))
            print()

            if total_nuevo > total:
                print("Aumentó la cantidad total de cable necesaria.")
            elif total_nuevo < total:
                print("Disminuyó la cantidad total de cable necesaria.")
            else:
                print("Se mantuvo igual la cantidad total de cable necesaria.")

            fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(22, 6))
            dibujar(componentes, cables=mst, titulo="Protoboard - MST original", ax=ax1)
            dibujar(componentes_nuevos, cables=mst_nuevo, titulo="Protoboard - MST después del cambio", ax=ax2)
            plt.show()

        elif opcion == 0:
            print("Muchas gracias por su tiempo, feliz dia, saliendo...")
            break

        else:
            print("Opción incorrecta, intentelo nuevamente.")

menu()
```
