## 1) Descripción del problema **[1 pts]**

Contexto
Una tienda necesita mejorar la forma en que consulta los productos de su inventario.

Actualmente tiene muchos productos registrados y recorrer toda una lista cada vez que necesita buscar uno resulta poco eficiente.

Para solucionar este problema se deberá implementar una Tabla Hash, utilizando como llave el código numérico del producto.

Cada producto tendrá:

Código Nombre Precio Cantidad disponible
Por ejemplo:

101 - Arroz       - $4.500  - 20 unidades 205 - Leche       - $3.800  - 15 unidades 312 - Café        - $12.000 - 10 unidades 423 - Pan         - $2.500  - 30 unidades
Punto 1. Crear la Tabla Hash
El usuario deberá indicar el tamaño m de la tabla.

Ejemplo:

Ingrese el tamaño de la tabla hash: 7
Inicialmente la tabla estará vacía:

Índice     Producto 0          Vacío 1          Vacío 2          Vacío 3          Vacío 4          Vacío 5          Vacío 6          Vacío
Para determinar la posición de cada producto se deberá utilizar:

h(código) = código % m
Donde:

código es la llave del producto.

m es el tamaño de la tabla.

Punto 2. Insertar productos
Permita al usuario registrar una cantidad n de productos.

Por ejemplo:

¿Cuántos productos desea registrar?: 5 Código: 101 Nombre: Arroz Precio: 4500 Cantidad: 20
Para cada producto deberá calcularse su posición utilizando la función hash.

Si:

m = 7 código = 101
entonces:

101 % 7 = 3
Por lo tanto, el producto deberá almacenarse inicialmente en la posición 3.

Después de insertar todos los productos, muestre el estado completo de la tabla.

Punto 3. Manejo de colisiones
Puede ocurrir que dos productos generen el mismo índice.

Por ejemplo:

Código 101: 101 % 7 = 3 Código 108: 108 % 7 = 3
Existe una colisión porque ambos productos intentan ocupar la posición 3.

Las colisiones deberán resolverse utilizando encadenamiento.

Por ejemplo:

Índice 3: 101 - Arroz      ↓ 108 - Huevos      ↓ 115 - Azúcar
Cada posición de la tabla podrá contener varios productos cuando exista una colisión.

Punto 4. Buscar un producto
El usuario deberá ingresar el código del producto que desea consultar.

Ejemplo:

Ingrese el código del producto: 108
El programa deberá:

Calcular su posición mediante la función hash.

Acceder directamente a esa posición.

Buscar el producto dentro de la posición correspondiente.

Si existe:

Producto encontrado Código: 108 Nombre: Huevos Precio: $16.000 Cantidad disponible: 12
Si no existe:

El producto no está registrado.
No se deberá recorrer toda la Tabla Hash para realizar la búsqueda.

Punto 5. Eliminar un producto
Permita eliminar un producto utilizando su código.

Ejemplo:

Código del producto que desea eliminar: 108
El programa deberá calcular su posición y eliminar únicamente ese producto.

Antes:

Índice 3: 101 - Arroz 108 - Huevos 115 - Azúcar
Después:

Índice 3: 101 - Arroz 115 - Azúcar
Si el código no existe, deberá mostrar:

No se encontró el producto.
Punto 6. Actualizar inventario
Permita buscar un producto por su código y modificar su cantidad disponible.

Ejemplo:

Código: 101 Producto: Arroz Cantidad actual: 20 Nueva cantidad: 35
Resultado:

Inventario actualizado correctamente. Arroz - 35 unidades
La búsqueda del producto deberá realizarse utilizando la Tabla Hash.

Punto 7. Mostrar la Tabla Hash
El programa deberá permitir visualizar el contenido completo de la estructura.

Por ejemplo:

========================================              TABLA HASH ======================================== 0 -> Vacío 1 -> 302 - Galletas 2 -> Vacío 3 -> 101 - Arroz      108 - Huevos      115 - Azúcar 4 -> 312 - Café 5 -> Vacío 6 -> 209 - Leche
Cuando existan varias entradas en una misma posición, deberán visualizarse todas para evidenciar las colisiones.

Caso de éxito: El usuario ingresa un codigo y con la funcion hash este es asignado.

**HU3: Manejo de colisiones en inserción**

Caso de éxito: El índice ya tiene un producto. El usuario ingresa un codigo y se asigna a un indice en el que ya existe un producto. El sistema almacena ambos productos en el índice 3 mediante encadenamiento.   

Caso de error: Falla la asignación de memoria para el nuevo nodo encadenado en el índice correspondiente, mostrando un error de sistema. (Nota: Condición de error técnica estándar).

Caso de éxito: El usuario ingresa un codigo. El sistema calcula su posición, lo busca en los elementos encadenados del índice y elimina únicamente ese producto, manteniendo los demás.

## 6) Código funcional **[15 pts]**
Incluye tu implementación completa.

No pegues código incompleto; debe ser ejecutable.


```python
def crear_tabla(m):
    tabla = []
    for i in range(m):
        tabla.append([])
    return tabla

def calcular_indice(codigo, m):
    return codigo % m

def buscar_producto(tabla, m, codigo):
    indice = calcular_indice(codigo, m)
    for producto in tabla[indice]:
        if producto["codigo"] == codigo:
            return producto
    return None

def insertar(tabla, m, codigo, nombre, precio, cantidad):
    if buscar_producto(tabla, m, codigo) != None:
        print("Error: Ya existe un producto con ese código.")
        return False
    indice = calcular_indice(codigo, m)
    producto = {"codigo": codigo, "nombre": nombre, "precio": precio, "cantidad": cantidad}
    tabla[indice].append(producto)
    return True

def buscar(tabla, m, codigo):
    producto = buscar_producto(tabla, m, codigo)
    if producto == None:
        print("El producto no está registrado.")
    else:
        precio = f"{producto['precio']:,}".replace(",", ".")
        print("Producto encontrado")
        print("Código:", producto["codigo"], "Nombre:", producto["nombre"], "Precio: $" + precio, "Cantidad disponible:", producto["cantidad"])

def eliminar(tabla, m, codigo):
    producto = buscar_producto(tabla, m, codigo)
    if producto == None:
        print("No se encontró el producto.")
    else:
        indice = calcular_indice(codigo, m)
        
        print("Antes:")
        cadena_antes = "Índice " + str(indice) + ":"
        for p in tabla[indice]:
            cadena_antes += " " + str(p["codigo"]) + " - " + p["nombre"]
        print(cadena_antes)
        
        tabla[indice].remove(producto)
        
        print("Después:")
        cadena_despues = "Índice " + str(indice) + ":"
        for p in tabla[indice]:
            cadena_despues += " " + str(p["codigo"]) + " - " + p["nombre"]
        print(cadena_despues)

def actualizar(tabla, m, codigo, nueva_cantidad):
    producto = buscar_producto(tabla, m, codigo)
    if producto == None:
        print("El producto no está registrado.")
    else:
        print("Código:", producto["codigo"], "Producto:", producto["nombre"], "Cantidad actual:", producto["cantidad"], "Nueva cantidad:", nueva_cantidad)
        producto["cantidad"] = nueva_cantidad
        print("Resultado:")
        print("Inventario actualizado correctamente.", producto["nombre"], "-", nueva_cantidad, "unidades")

def mostrar_tabla(tabla):
    print("TABLA HASH")
    for i in range(len(tabla)):
        if len(tabla[i]) == 0:
            print(i, "-> Vacío")
        else:
            cadena = str(i) + " ->"
            for producto in tabla[i]:
                cadena += " " + str(producto["codigo"]) + " - " + producto["nombre"]
            print(cadena)

def pedir_entero(mensaje, mayor_cero):
    while True:
        try:
            valor = int(input(mensaje))
            if mayor_cero and valor <= 0:
                print("El valor debe ser mayor que 0.")
            elif not mayor_cero and valor < 0:
                print("El valor no puede ser negativo.")
            else:
                return valor
        except ValueError:
            print("Entrada inválida. Por favor, ingrese un número entero.")

def menu():
    print("Bienvenido se realizara una gestion de su inventario para mayor facilidad e interactividad con sus productos.")
    m = pedir_entero("Ingrese el tamaño de la tabla hash: ", True)

    tabla = crear_tabla(m)
    mostrar_tabla(tabla)

    opcion = ""
    while opcion != "6":
        print("\tMenu del inventario")
        print("1. Insertar productos")
        print("2. Buscar un producto")
        print("3. Eliminar un producto")
        print("4. Actualizar cantidad")
        print("5. Mostrar la tabla Hash")
        print("6. Salir")
        opcion = input("Seleccione una opción: ")

        if opcion == "1":
            n = pedir_entero("¿Cuántos productos desea registrar?: ", True)
            registrados = 0
            while registrados < n:
                codigo = pedir_entero("Código: ", False)
                nombre = input("Nombre: ")
                precio = pedir_entero("Precio: ", False)
                cantidad = pedir_entero("Cantidad: ", False)
                
                if insertar(tabla, m, codigo, nombre, precio, cantidad):
                    registrados += 1
            mostrar_tabla(tabla)

        elif opcion == "2":
            codigo = pedir_entero("Ingrese el código del producto: ", False)
            buscar(tabla, m, codigo)

        elif opcion == "3":
            codigo = pedir_entero("Código del producto que desea eliminar: ", False)
            eliminar(tabla, m, codigo)

        elif opcion == "4":
            codigo = pedir_entero("Código: ", False)
            producto = buscar_producto(tabla, m, codigo)
            if producto == None:
                print("El producto no está registrado.")
            else:
                nueva_cantidad = pedir_entero("Nueva cantidad: ", False)
                actualizar(tabla, m, codigo, nueva_cantidad)

        elif opcion == "5":
            mostrar_tabla(tabla)

        elif opcion == "6":
            print("Muchas gracias por su tiempo feliz dia...")

        else:
            print("Opción inválida.")

menu()
```
