## Enunciado

**Ejercicio: Organización de precios en una tienda**

Una tienda necesita organizar los precios de sus productos. La cantidad de productos puede variar, por lo que el usuario deberá indicar cuántos precios desea registrar.

El programa debe solicitar:
1. La cantidad de productos, representada por n.
2. El precio de cada uno de los productos.
3. El tipo de ordenamiento que se desea mostrar.

Después de ingresar los datos, el programa deberá utilizar el algoritmo **Insertion Sort** para organizar los precios.

**Punto 1. Orden ascendente**

Ordenar los precios desde el menor hasta el mayor, con el fin de mostrar primero los productos más económicos.

**Punto 2. Orden descendente**

Ordenar los precios desde el mayor hasta el menor, con el fin de mostrar primero los productos más costosos.

# Especificacion




**Entradas**

1. La primera entrada que se tendra es la cantidad de productos a ordenar (tiene que ser mayor que 1).
2. Todos los precios de los productos que el usuario desea ordenar (sin duplicados)
3. Todos los nombres de los productos que el usuario desea ordenar (mayores a 0).
4. Tipo de ordenamiento que desea que se realice, ya sea de menor a mayor o de mayor a menor (solo como respuesta "ASCENDENTE" o "DESCENDENTE").

En resumen, el programa recibe la cantidad de productos a comparar, solicita el nombre y el valor de cada producto y despues solicita la manera en la que se desea que se organicen los productos, si de manera ascendente o descendente.

**Salida**

Como salida exitosa, se tiene una permutacion de los productos y sus precios ordenados segun el tipo de ordenamiento que escogio el usuario.

En cuanto a salidas de error, se le entrega su correspondiente salida en caso de que las entradas sean invalidas.

**Restricciones**

1. Los precios deben tener un valor mayor a 0.
2. Se deben registrar al menos 2 elementos.
3. Se debe escoger uno de los 2 modelos de ordenamiento definidos.
4. Elegir cuando finalizar el programa.
5. Los productos no pueden repetirse

**Requerimientos Funcionales**

Para hacer este programa posible, son necesarios los siguientes requerimientos:

*   **RQF01**: Solicitar y validar la cantidad de productos que se quieren ordenar

*   **RQF06**: Reiniciar o finalizar el programa, depende de la respuesta del usuario

# Codigo


```python
def bienvenida():
    print() 
    print("\t\tBIENVENIDO AL PROGRAMA")
    print("Este programa le pedira la catidad de productos que desea ordenar \nasi mismo, se le pedira el nombre y precio de cada producto \nfinalmente, se le pedira que escoga una manera de organizar los prodcutos.")

def solicitar_cantidad():
    print("Por favor digite la cantidad de productos que quiere ordenar:")
    x = int(input())
    while x <= 1:
        print("-La cantidad de productos no puede ser menor o igual 1.")
        print("-Digite nuevamente:")
        x = int(input())
    return x 

def llenar_nombres_precios(cant):
    productos = []
    precios = []
    for rep in range (cant):
        print("Escriba el nombre del producto numero",rep+1,":")
        nombre_p = input()
        while nombre_p in productos:
            print("¡¡El producto ya se encuentra registrado, registre uno nuevo!!")
            nombre_p = input()
        print("Escriba el precio del prodcuto numero", rep+1,":")
        precio_p = float(input())
        while precio_p <= 0:
            print("Precio invalido, por favor digite nuevamente.")
            precio_p = float(input())    
        productos.append(nombre_p)
        precios.append(precio_p)
    return productos, precios

def solicitar_y_ordenar(cant, productos, precios):
    print("Si desea ordenar de mayor a menor, por favor escriba <DESCENDENTE> ")
    print("Si desea ordenar de menor a mayor, por favor escriba <ASCENDENTE> ")
    orden = input()
    while orden != "DESCENDENTE" and orden!="ASCENDENTE":
        print("Opcion invalida, por favor escriba nuevamente.")
        orden = input()
    if orden == "DESCENDENTE":
        for i in range(cant):
            for j in range(cant-1):
                if precios[j] < precios[j+1]:
                    precios[j], precios[j + 1] = precios[j + 1], precios[j]
                    productos[j], productos[j + 1] = productos[j + 1], productos[j]
    else:
        for i in range(cant):
            for j in range(cant-1):
                if precios[j] > precios[j+1]:
                    precios[j], precios[j + 1] = precios[j + 1], precios[j]
                    productos[j], productos[j + 1] = productos[j + 1], productos[j]

def entrega_resultados(cant, productos, precios):
    print("\tLISTADO DE PRODUCTOS \nPRODUCTOS \t PRECIOS")
    for i in range(cant):
        print(productos[i],"\t",precios[i])

def main():
    band = True
    bienvenida()
    while band:
        cant = solicitar_cantidad()
        productos, precios = llenar_nombres_precios(cant)
        solicitar_y_ordenar(cant, productos, precios)
        entrega_resultados(cant, productos, precios)
        print("Desea realizar el proceso nuevamente, Responder solo <SI> o <NO>.")
        x = input()
        while x != "SI" and x != "NO":
            print("Respuesta invalida, digite nuevamente.")
            x = input()
        if x == "NO":
            band = False
    print("Gracias por usar el programa.")

main()
```
